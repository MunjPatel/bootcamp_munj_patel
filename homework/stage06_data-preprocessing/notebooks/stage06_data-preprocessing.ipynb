### Pre-processing and cleaning the data

```python
import sys
import os

src_path = os.path.abspath(os.path.join('..', 'src'))

if src_path not in sys.path:
    sys.path.append(src_path)
```

```python
from cleaning import CleanData
```

```python
# instanciating the class

clean = CleanData()
```

```python
# missing value imputation

no_missing_data = clean.fill_missing_median(data_frame=df)
no_missing_data
```

```python
# drop missing values

drop_missing_data = clean.drop_missing(no_missing_data)
drop_missing_data
```

```python
# normalize the data

normalized_clean_data = clean.normalize_data(data_frame=drop_missing_data)
```

```python
normalized_clean_data_columns = drop_missing_data.select_dtypes(np.number).columns.tolist()

for index in range(normalized_clean_data.shape[1]):
    drop_missing_data[normalized_clean_data_columns[index]] = normalized_clean_data[:,index]
```

```python
# Save to CSV in raw data folder
processed_path = os.path.join(processed_dir, 'processed_data.csv')
if not os.path.exists(processed_path):
    drop_missing_data.to_csv(processed_path, index=False)
    print(f'Processed dataset created and saved to {processed_path}')
else:
    print(f'File already exists at {processed_path}. Skipping CSV creation to avoid overwrite.')
```
